```python
# 安装 scikit-fuzzy 库 (如果你在本地运行，请在终端输入 pip install scikit-fuzzy)
!pip install scikit-fuzzy

import numpy as np
import skfuzzy as fuzz
from skfuzzy import control as ctrl
import matplotlib.pyplot as plt
```

```python
# 1. 定义论域 (Universe of Discourse)
# np.arange(起点, 终点+步长, 步长)
severity = ctrl.Antecedent(np.arange(0, 11, 1), 'severity')           # 症状严重程度：0到10
pain = ctrl.Antecedent(np.arange(0, 11, 1), 'pain')                   # 疼痛等级：0到10
vulnerability = ctrl.Antecedent(np.arange(0, 101, 1), 'vulnerability')# 脆弱度(年龄/基础病)：0到100

priority = ctrl.Consequent(np.arange(0, 101, 1), 'priority')          # 优先级：0到100

# 2. 手动定义模糊集合 (隶属函数)
# 格式：fuzz.trimf(论域, [左边界, 顶点, 右边界])

# 症状严重程度：轻微(mild), 中等(moderate), 严重(severe)
severity['mild'] = fuzz.trimf(severity.universe, [0, 0, 5])
severity['moderate'] = fuzz.trimf(severity.universe, [2, 5, 8])
severity['severe'] = fuzz.trimf(severity.universe, [5, 10, 10])

# 疼痛等级：低(low), 中(medium), 高(high)
pain['low'] = fuzz.trimf(pain.universe, [0, 0, 5])
pain['medium'] = fuzz.trimf(pain.universe, [2, 5, 8])
pain['high'] = fuzz.trimf(pain.universe, [5, 10, 10])

# 脆弱度风险：低(low), 中(medium), 高(high)
vulnerability['low'] = fuzz.trimf(vulnerability.universe, [0, 0, 50])
vulnerability['medium'] = fuzz.trimf(vulnerability.universe, [25, 50, 75])
vulnerability['high'] = fuzz.trimf(vulnerability.universe, [50, 100, 100])

# 输出 - 分诊优先级：常规(routine), 紧急(urgent), 抢救(emergency)
priority['routine'] = fuzz.trimf(priority.universe, [0, 0, 50])
priority['urgent'] = fuzz.trimf(priority.universe, [25, 50, 75])
priority['emergency'] = fuzz.trimf(priority.universe, [50, 100, 100])

# 【截图点】查看隶属函数图 (报告需要)
severity.view()
priority.view()
```

```python
# 定义 11 条模糊规则 (使用 & 代表 AND, | 代表 OR)
rule1 = ctrl.Rule(severity['severe'] & pain['high'], priority['emergency'])
rule2 = ctrl.Rule(severity['mild'] & vulnerability['low'], priority['routine'])
rule3 = ctrl.Rule(severity['moderate'] & vulnerability['high'], priority['urgent'])
rule4 = ctrl.Rule(severity['severe'] & pain['low'], priority['urgent'])
rule5 = ctrl.Rule(severity['mild'] & pain['high'], priority['urgent'])
rule6 = ctrl.Rule(vulnerability['high'] & pain['high'], priority['emergency'])
rule7 = ctrl.Rule(severity['moderate'] & pain['medium'], priority['urgent'])
rule8 = ctrl.Rule(severity['mild'] & vulnerability['medium'], priority['routine'])
rule9 = ctrl.Rule(severity['moderate'] & vulnerability['low'], priority['routine'])
rule10 = ctrl.Rule(severity['severe'] & vulnerability['high'], priority['emergency'])
rule11 = ctrl.Rule(severity['mild'] & pain['low'], priority['routine'])

# 如果你想在报告里展示规则，可以取消下面这行的注释
# rule1.view()
```

```python
# 创建控制系统
triage_ctrl = ctrl.ControlSystem([rule1, rule2, rule3, rule4, rule5, rule6, rule7, rule8, rule9, rule10, rule11])

# 创建运行模拟器
triage_sim = ctrl.ControlSystemSimulation(triage_ctrl)
```

```python
# 假设送来一位患者：
# 症状严重程度评分为 8 (比较严重)
# 疼痛等级为 7 (比较痛)
# 脆弱度指数为 80 (比如是一位 80 岁的老人)

triage_sim.input['severity'] = 8
triage_sim.input['pain'] = 7
triage_sim.input['vulnerability'] = 80

# 执行计算 (去模糊化)
triage_sim.compute()

# 打印最终的优先级分数 (Crisp Output)
print(f"推荐的分诊优先级得分为: {triage_sim.output['priority']:.2f} / 100")

# 【截图点】可视化最终输出结果 (质心去模糊化图)
priority.view(sim=triage_sim)
```
